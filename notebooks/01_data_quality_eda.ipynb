```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
jobs = pd.read_csv(ROOT / 'data/processed/jobs_clean.csv')
quality = json.loads((ROOT / 'reports/data_quality.json').read_text(encoding='utf-8'))
quality
```

```python
missing = jobs.isna().mean().mul(100).sort_values(ascending=False)
display(missing.to_frame('missing_percent'))
missing[missing.gt(0)].sort_values().plot.barh(figsize=(8, 5), title='Missing values (%)')
plt.tight_layout()
```

```python
display(jobs['posisi'].value_counts().head(15).to_frame('vacancies'))
display(jobs['provinsi'].value_counts(dropna=False).head(15).to_frame('vacancies'))
salary = jobs['salary_monthly_idr'].dropna()
display(salary.describe().to_frame('monthly_salary_idr'))
salary.plot.hist(bins=15, title='Disclosed monthly salaries');
```

Salary exists in only 13.9% of rows. These results describe this sample, not the entire Indonesian labour market. Without a validated posting date, the data cannot support time-series trend claims.
